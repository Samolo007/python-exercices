```python
import seaborn as sns
import pandas as pd
df = pd.read_csv('../exercice-08/data/titanic_propre.csv')
```

1-les femmes ont-elles mieux survécu que les hommes dans chaque classe?

```python
df.groupby("pclass")["survived"].mean()
resultat = df.groupby(["pclass", "sex"]).agg(
taux_survie=("survived", "mean"),
).round(2)
print(resultat)
```

Oui, les femmes ont mieux survécu que les hommes dans chaque classe selon les résultats du taux de survie.

2-Les enfants ont-ils été prioritaires? Taux de survie par tranche d'âge

```python
df['Tranche_Age'] = pd.cut(df['age'], bins=[0, 12, 18, 35, 60, 100], labels=['Enfants', 'Adolescents', 'Adultes jeunes', 'Adultes', 'Seniors'], right=False)
resultat = df.groupby("Tranche_Age", observed=False).agg(
    passagers=("survived", "size"),
    taux_survie=("survived", "mean"))
print(resultat)
```

Oui, les enfants ont été prioritaires avec le taux de survie le plus élevé.

3-Quel port a le prix moyen le plus élevé? La répartition des classes l'explique-t-elle?

```python
prix_moyen_port = df.groupby("embark_town")["fare"].mean()
print("Prix moyen par port :")
print(prix_moyen_port)
```

```python
# Répartition des classes par port en pourcentage
repartition = pd.crosstab(df["embark_town"], df["pclass"], normalize="index") * 100
print(repartition)
```

Le port moyen le plus élevé est Cherbourg; cela s'explique par son pourcentage élevé de voyageurs en première classe.

4-Voyager seul augmentait-il ou diminuait-il les chances de survie?

```python
df["is_alone"] = (df["sibsp"] + df["parch"]) == 0
taux_survie_seul = df.groupby("is_alone")["survived"].mean() * 100
print(taux_survie_seul)
```

Selon le résultat obtenu, voyager seul diminue les chances de survie.
