```python
import seaborn as sns
titanic = sns.load_dataset("titanic")   
import pandas as pd
```

1-Nombre de passagers mineurs

```python
nombre_de_passagers = (titanic["age"] < 18).sum()
print(nombre_de_passagers)
```

2- les 5 billets les plus chers

```python
titanic.sort_values("fare", ascending=False)[["fare", "pclass", "embarked"]].head(5)
```

3-pourcentage des femmes en première classe

```python
femmes_première_classe = titanic[(titanic["sex"]== "female") & (titanic["pclass"] == 1)]
pourcentage = (len(femmes_première_classe)/ len(titanic[titanic["pclass"] ==1]))* 100
print(pourcentage)
```

4-taux de survie des femmes en première classe et des hommes en 3ème classe

```python
taux_survie_femmes = femmes_première_classe["survived"].mean()*100
print(taux_survie_femmes)
```

```python
hommes_troisième_classe = titanic[(titanic["sex"]== "male") & (titanic["pclass"] == 3)]
taux_survie_hommes = hommes_troisième_classe["survived"].mean()*100
print(taux_survie_hommes)
```
