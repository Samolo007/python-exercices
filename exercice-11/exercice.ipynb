```python
import pandas as pd
import seaborn as sns
taxis = sns.load_dataset("taxis") 
```

1-

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
taxis.shape
```

2-

```python
ca_par_jour = (
    taxis.set_index("pickup")
    .resample("D")
    .agg(
        nombre_courses=(
            "total",
            "count",
        ),
      chiffre_affaires=("total", "sum"),
      )
    .reset_index()
)
print(ca_par_jour.head())
```

3-

```python
taxis["heure"] = taxis["pickup"].dt.hour
heure_plus_chargee = (
    taxis.groupby("heure").size().reset_index(name="nombre_courses")
)
heure_plus_chargee = heure_plus_chargee.sort_values(
    by="nombre_courses", ascending=False
)
print(heure_plus_chargee.head())
```

4-

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()
pourboire_par_jour = (
    taxis.groupby("jour_semaine")["tip"].mean().reset_index()
)
pourboire_par_jour = pourboire_par_jour.sort_values(
    by="tip", ascending=False
)

print(pourboire_par_jour)
```

5-

```python
taxis["duree_minutes"] = (
    taxis["dropoff"] - taxis["pickup"]
).dt.total_seconds() / 60
print(taxis["duree_minutes"].describe())
```
