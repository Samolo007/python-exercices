```python
import seaborn as sns
import pandas as pd
df = pd.read_csv('../exercice-08/data/titanic_propre.csv')
```

1-

```python
data = {
    "embark_town": ['Southampton', 'Cherbourg', 'Queenstown'],
    "Pays": ['Angleterre', 'France', 'Irlande']
}
ports = pd.DataFrame(data)
ports
```

2-

```python
df = df.merge(ports, on="embark_town", how="left")
len(df)
```

3-

```python
taux_par_pays = df.groupby('Pays')['survived'].agg(
    Nombre_passagers='count',
    Taux_survie='mean'
)
print(taux_par_pays)
```

4-

```python
data_2 = {
    "embark_town": ['Southampton', 'Cherbourg'],
    "Pays": ['Angleterre', 'France']
}
ports = pd.DataFrame(data_2)
```

```python
df_inner = df.merge(ports, on="embark_town", how="inner")
print(len(df_inner))
```

```python
df_left = df.merge(ports, on="embark_town", how="left")
print( len(df_left))
```

Quand on retire Queenstown de port et qu'on fait la jointure en "inner", toutes les lignes qui était relié à Queenstown sont supprimés. Quand on fait une jointure "left", ces lignes reapparaissent.
