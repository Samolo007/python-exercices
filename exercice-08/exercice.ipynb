```python
import seaborn as sns
titanic = sns.load_dataset("titanic")   
import pandas as pd
```

1-

```python
titanic["age"] = titanic["age"].fillna(
    titanic.groupby(["class", "sex"], observed=True)["age"].transform("median")
)
titanic = titanic.drop(columns=["deck"], errors="ignore")
titanic = titanic.dropna(subset=["embark_town"])
print(titanic["age"])
print(titanic)
```

2-

```python
print(titanic.duplicated().sum())
doublons = titanic[titanic.duplicated(keep=False)]
print(doublons)
```

3-

```python
print((titanic["survived"].map({0: "no", 1: "yes"}) == titanic["alive"]).all())  
print(pd.crosstab(titanic["pclass"], titanic["class"]))          
print(pd.crosstab(titanic["embarked"], titanic["embark_town"]))  
print((titanic["alone"] == ((titanic["sibsp"] + titanic["parch"]) == 0)).all())  
print((titanic["adult_male"] == (titanic["who"] == "man")).all())                
```

```python
titanic["survived"] = titanic["survived"].astype(bool)
titanic["pclass"] = pd.Categorical(titanic["pclass"], categories=[1, 2, 3], ordered=True)
for col in ["sex", "who", "embark_town"]:
    titanic[col] = titanic[col].astype("category")
titanic[["sibsp", "parch"]] = titanic[["sibsp", "parch"]].astype("int8")

titanic.info()
```
