```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
df = pd.read_csv('../exercice-08/data/titanic_propre.csv')
taxis = sns.load_dataset("taxis") 
```

1-

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    df["age"].dropna(),
    kde=True,
    bins=30,
    color="skyblue",
    edgecolor="black",
)
plt.title("Distribution de l'âge des passagers du Titanic")
plt.xlabel("Âge")
plt.ylabel("Nombre de passagers")
os.makedirs("figures", exist_ok=True)
plt.savefig("figures/age_passagers.png", bbox_inches="tight")
plt.show()
```

JUSTIFICATION : ce diagramme permet de représenter l'évolution d'une variable(âge des passagers) en fonction d'une autre(Nombre de passagers). 
CONCLUSION : l'âge moyenne des passagers étaient entre 20 et 30 ans.

2-

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=df, x="pclass", y="survived", hue="sex", palette="Set2", ax=ax)
ax.set_title("Taux de survie selon la classe et le sexe")
ax.set_xlabel("Classe")
ax.set_ylabel("Taux de survie")
ax.legend(title="Sexe")
fig.savefig("figures/survie_classe_sexe.png", dpi=150, bbox_inches="tight")
plt.show()

```

JUSTIFICATION : ce diagramme permet de comparer les valeurs discrètes entre plusieurs catégories. 
CONCLUSION : Selon la classe et le sexe, on peut remarquer que les femmes ont plus survécu que les hommes.

3-

```python
os.makedirs("figures", exist_ok=True)
fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(
    data=df,
    x="pclass",
    y="fare",
    hue="pclass",
    palette="Set3",
    legend=False,
    ax=ax,
)
ax.set_title("Distribution du prix du billet selon la classe")
ax.set_xlabel("Classe")
ax.set_ylabel("Prix du billet")
fig.savefig("figures/prix_selon_classe.png", dpi=150, bbox_inches="tight")
plt.show()
```

JUSTIFICATION : ce diagramme permet de comparer la distribution d'une variable quantitative (prix du billet) en fonction d'une variable qualitative (classe)
CONCLUSION : la première classe affiche des prix élevées par rapport aux autres classes.

4-

```python
ca_par_jour = (
    taxis.set_index("pickup")
    .resample("D")
    .agg(nombre_courses=("total", "count"), chiffre_affaires=("total", "sum"))
    .reset_index()
)
fig, ax = plt.subplots(figsize=(12, 5))

sns.lineplot(
    data=ca_par_jour,
    x="pickup",
    y="chiffre_affaires",
    color="crimson",
    ax=ax,
)

ax.set_title("Évolution du chiffre d'affaires des taxis jour après jour")
ax.set_xlabel("Date")
ax.set_ylabel("Chiffre d'affaires ($)")
fig.savefig("figures/ca_taxis_par_jour.png", dpi=150, bbox_inches="tight")
plt.show()
```

JUSTIFICATION : ce diagramme nous montre la variation d'une donnée au fil du temps
CONCLUSION : le chiffre d'affaire a atteint son max en 2019-03-13. il a connu des evolutions et des regressions au fil du temps.

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(
    data=taxis, x="distance", y="total", alpha=0.5, color="purple", ax=ax
)
ax.set_title("Lien entre la distance de la course et le prix total")
ax.set_xlabel("Distance (miles)")
fig.savefig("figures/corrélation_distance_course", dpi=150, bbox_inches="tight")
plt.show()
```

JUSTIFICATION : ce diagramme permet d'analyser la corrélation entre deux variables quantitatives continues (distance et prix)
CONCLUSION : Comme on peut le voir, il existe une corrélation entre ces deux variables. Plus la distance augmente, plus on voit une augmentation du prix
